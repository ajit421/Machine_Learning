```python
import numpy as np
import pandas as pd
```

```python
Data = {
    'feature1': [1, 2, np.nan, 4, 5],
    'feature2': [np.nan, 4, 6, 8, 3],
    'feature3': [3, 0, 1, 1, np.nan]
}

df = pd.DataFrame(Data)
df
```

```python
from sklearn.impute import KNNImputer
imputer = KNNImputer()
df_imputed = imputer.fit_transform(df)
df_imputed
```
