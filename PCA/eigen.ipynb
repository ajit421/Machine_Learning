```python
import numpy as np
```

```python
x = np.array([
    [-1, 1, 3],
    [1, 2, 0],
    [3, 0, 2]
]) 
```

```python
eigenvalues, eigenvectors = np.linalg.eig(x)
```

```python
print("eigenvalues", eigenvalues)
print("eigenvectors\n", eigenvectors)
```

2D to 1D

https://shad.io/MatVis/

```python
import matplotlib.pyplot as plt

fig = plt.figure(figsize=(10, 8))
ax = fig.add_subplot(111, projection='3d')

# Plot original vectors (columns of x)
origin = [0, 0, 0]
for i in range(x.shape[1]):
    ax.quiver(*origin, x[0, i], x[1, i], x[2, i], color='blue', alpha=0.5, label='Original Vector' if i == 0 else "")

# Plot eigenvectors
for i in range(eigenvectors.shape[1]):
    # Scale by eigenvalue for visualization
    vec = eigenvectors[:, i] * eigenvalues[i]
    ax.quiver(*origin, vec[0], vec[1], vec[2], color='red', label='Eigenvector' if i == 0 else "")

ax.set_xlim([-5, 5])
ax.set_ylim([-5, 5])
ax.set_zlim([-5, 5])
ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('Z')
ax.legend()
plt.title('3D Visualization of Vectors and Eigenvectors')
plt.show()
```
